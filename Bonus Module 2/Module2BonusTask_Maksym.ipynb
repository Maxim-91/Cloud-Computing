# From Huggingface hub: distilbert-base-uncased-finetuned-sst-2-english

```python
classifier = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)
```

# 1. Typo testing

```python
typo_tests = [
    "I love this movie",
    "I loooove this movie",
    "I lvoe this movie",
    "I lovee this moovie",
    "Ths movie is gr8"
]

for text in typo_tests:
    print(text, "-", classifier(text))
```

# 2. Slang testing

```python
slang_tests = [
    "This movie slaps",
    "This movie is fire",
    "That was meh",
    "This movie sucks",
    "This show is lit"
]

for text in slang_tests:
    print(text, "-", classifier(text))
```

# 3. Irony testing

```python
irony_tests = [
    "Oh great, another Monday. Just what I needed.",
    "Fantastic service… not!",
    "I’m so happy I failed my exam",
    "Yeah, that was the best meal ever… not.",
    "I absolutely love waiting in traffic for hours"
]

for text in irony_tests:
    print(text, "-", classifier(text))
```
